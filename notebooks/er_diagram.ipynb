```python
import pandas as pd
import glob, os, itertools
```

```python
DATA_DIR = "../data"
dfs = {os.path.splitext(os.path.basename(f))[0]: pd.read_csv(f, low_memory=False)
       for f in glob.glob(f"{DATA_DIR}/*.csv")}
```

```python
for n, d in dfs.items():
    print(f"\n[{n}]  rows={len(d):,}")
    print("  cols:", ", ".join(d.columns))

# 2) relationships
def kind(u1, u2):
    return {(True, True): "1 : 1", (True, False): "1 : many",
            (False, True): "many : 1", (False, False): "many : many"}[(u1, u2)]

print("\n=== RELATIONSHIPS ===")
for (a, da), (b, db) in itertools.combinations(dfs.items(), 2):
    for col in sorted(set(da.columns) & set(db.columns)):
        sa, sb = da[col].dropna(), db[col].dropna()
        if sa.empty or sb.empty:
            continue
        shared = len(set(sa) & set(sb))
        pct = shared / min(sa.nunique(), sb.nunique())
        if pct >= 0.5:
            print(f"{a}.{col}  --{kind(sa.is_unique, sb.is_unique)}-->  {b}.{col}"
                  f"   (shared values: {shared:,}, {pct:.0%} overlap)")
```
